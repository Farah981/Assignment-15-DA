1. List all cities where the area is greater than 5000, along with their population.

```python
pd.read_sql_query('''
select  c.name City_name , c.population Population ,f.area Area_name from cities c
 inner join facts f on c.facts_id =  f.id and area > 5000
''',conn)
```

2. List the names of capital cities where the death rate is greater than 2% and the
birth rate is less than 20%

```python
pd.read_sql_query('''
select f.death_rate , f.birth_rate,c.name as Capital_City from cities c 
inner join facts f on f.id = c.facts_id where f.death_rate >2 and f.birth_rate <20

    
''',conn)
```

3. List the cities with a population growth rate between 1% and 5%, sorted in desc
by population growth.

```python
pd.read_sql_query('''
select c.name Cities_Name ,f.population_growth Population_Growth from cities c
inner join facts f on c.facts_id = f.id where f.population_growth between 1 and 5 order by 2 desc
    
''',conn)
```

4. Find countries that start with &quot;A&quot; and do not have any cities listed in the cities
table. The query should return the country name and city name. Using left join

```python
pd.read_sql_query('''
SELECT f.name AS Country_Name,
       c.name AS City_Name
FROM facts f
LEFT JOIN cities c
ON f.id = c.facts_id
WHERE f.name LIKE 'A%'
AND c.id IS NULL
''', conn)
```

5. List all capital cities and their corresponding death rate using a LEFT JOIN.

```python
pd.read_sql_query('''
SELECT c.name AS Capital_City,
       f.death_rate AS Death_Rate
FROM cities c
LEFT JOIN facts f
ON c.facts_id = f.id
WHERE c.capital = 1
''', conn)
```
